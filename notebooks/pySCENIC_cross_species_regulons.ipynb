For a homologous TF pair, the workflow can:
- compare final motif-supported target sets;
- calculate direct or orthology-aware target overlap;
- rank a regulon's targets by how preferentially they appear among highly

The goal is to compare regulatory programs without assuming that identical
gene symbols imply identical cross-species targets.
